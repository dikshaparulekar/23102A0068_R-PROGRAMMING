```r
# ============================================================
# LAB 6: Statistical Analysis of Palmer Penguins (LIGHTWEIGHT VERSION)
# No car / rstatix / ggpubr — avoids slow dependency installs
# ============================================================

# ---- 1. Install only light packages, with dependencies = FALSE ----
required_packages <- c("palmerpenguins", "dplyr", "ggplot2", "moments")

for (p in required_packages) {
  if (!requireNamespace(p, quietly = TRUE)) {
    install.packages(p, dependencies = FALSE)
  }
}

library(palmerpenguins)
library(dplyr)
library(ggplot2)
library(moments)   # skewness, kurtosis

# ---- 2. Load & Clean Dataset ----
data("penguins")
df <- penguins %>% na.omit()

cat("Dataset dimensions:", dim(df), "\n")
str(df)

# ============================================================
# TASK 1: DESCRIPTIVE STATISTICAL ANALYSIS (body_mass_g)
# ============================================================

cat("\n========== DESCRIPTIVE STATISTICS: BODY MASS (Overall) ==========\n")

desc_stats <- function(x) {
  c(Mean = mean(x), Median = median(x), Min = min(x), Max = max(x),
    Variance = var(x), SD = sd(x),
    Q1 = quantile(x, 0.25), Q3 = quantile(x, 0.75), IQR = IQR(x),
    Skewness = skewness(x), Kurtosis = kurtosis(x))
}
print(round(desc_stats(df$body_mass_g), 3))

cat("\n========== DESCRIPTIVE STATISTICS: BODY MASS (By Species) ==========\n")
species_desc <- df %>%
  group_by(species) %>%
  summarise(Mean = mean(body_mass_g), Median = median(body_mass_g),
            Min = min(body_mass_g), Max = max(body_mass_g),
            Variance = var(body_mass_g), SD = sd(body_mass_g),
            Q1 = quantile(body_mass_g, 0.25), Q3 = quantile(body_mass_g, 0.75),
            IQR = IQR(body_mass_g), Skewness = skewness(body_mass_g),
            Kurtosis = kurtosis(body_mass_g))
print(species_desc)

# --- Visualizations ---
ggplot(df, aes(x = body_mass_g)) +
  geom_histogram(binwidth = 100, fill = "steelblue", color = "black") +
  labs(title = "Histogram of Body Mass", x = "Body Mass (g)", y = "Frequency") +
  theme_minimal()
ggsave("hist_body_mass.png", width = 7, height = 5)

ggplot(df, aes(x = species, y = body_mass_g, fill = species)) +
  geom_boxplot() +
  labs(title = "Species-wise Boxplot of Body Mass", x = "Species", y = "Body Mass (g)") +
  theme_minimal()
ggsave("boxplot_species_body_mass.png", width = 7, height = 5)

ggplot(df, aes(x = sex, y = body_mass_g, fill = sex)) +
  geom_boxplot() +
  labs(title = "Sex-wise Boxplot of Body Mass", x = "Sex", y = "Body Mass (g)") +
  theme_minimal()
ggsave("boxplot_sex_body_mass.png", width = 7, height = 5)

ggplot(df, aes(x = body_mass_g, fill = species)) +
  geom_density(alpha = 0.5) +
  labs(title = "Density Plot of Body Mass by Species", x = "Body Mass (g)", y = "Density") +
  theme_minimal()
ggsave("density_body_mass.png", width = 7, height = 5)

# ============================================================
# TASK 2: HYPOTHESIS TESTING (Male vs Female Body Mass)
# ============================================================

cat("\n========== HYPOTHESIS TESTING: MALE vs FEMALE BODY MASS ==========\n")
cat("H0: Mean body mass (male) = Mean body mass (female)\n")
cat("H1: Mean body mass (male) != Mean body mass (female)\n\n")

male_mass <- df$body_mass_g[df$sex == "male"]
female_mass <- df$body_mass_g[df$sex == "female"]

cat("Shapiro-Wilk Test (Male):\n"); print(shapiro.test(male_mass))
cat("\nShapiro-Wilk Test (Female):\n"); print(shapiro.test(female_mass))

png("qqplot_male_female.png", width = 800, height = 400)
par(mfrow = c(1, 2))
qqnorm(male_mass, main = "QQ Plot - Male"); qqline(male_mass, col = "red")
qqnorm(female_mass, main = "QQ Plot - Female"); qqline(female_mass, col = "red")
dev.off()
par(mfrow = c(1, 1))

t_test_result <- t.test(body_mass_g ~ sex, data = df, var.equal = FALSE)
cat("\nIndependent Two-Sample t-test:\n"); print(t_test_result)
cat("\n95% CI for difference in means:", t_test_result$conf.int, "\n")

# Manual Cohen's d (no 'effsize' package needed)
cohen_d <- function(x, y) {
  nx <- length(x); ny <- length(y)
  pooled_sd <- sqrt(((nx - 1) * var(x) + (ny - 1) * var(y)) / (nx + ny - 2))
  (mean(x) - mean(y)) / pooled_sd
}
d_val <- cohen_d(male_mass, female_mass)
cat("\nCohen's d Effect Size:", round(d_val, 3), "\n")
cat("Interpretation: p-value =", t_test_result$p.value,
    "-> ", ifelse(t_test_result$p.value < 0.05,
                   "Statistically significant difference.",
                   "No statistically significant difference."), "\n")

# ============================================================
# TASK 3: ONE-WAY ANOVA (body_mass_g ~ species)
# ============================================================

cat("\n========== ONE-WAY ANOVA: BODY MASS ~ SPECIES ==========\n")

cat("Shapiro-Wilk Test per Species:\n")
df %>% group_by(species) %>%
  summarise(p_value = shapiro.test(body_mass_g)$p.value) %>% print()

png("qqplot_species.png", width = 1200, height = 400)
par(mfrow = c(1, 3))
for (sp in levels(df$species)) {
  vals <- df$body_mass_g[df$species == sp]
  qqnorm(vals, main = paste("QQ Plot -", sp)); qqline(vals, col = "red")
}
dev.off()
par(mfrow = c(1, 1))

# Manual Levene's Test (Brown-Forsythe, median-based) - no 'car' package needed
manual_levene <- function(values, groups) {
  d <- data.frame(values = values, groups = groups)
  d <- d %>% group_by(groups) %>% mutate(abs_dev = abs(values - median(values)))
  aov(abs_dev ~ groups, data = d)
}
cat("\nLevene's Test (Brown-Forsythe, manual):\n")
levene_model <- manual_levene(df$body_mass_g, df$species)
print(summary(levene_model))

anova_model <- aov(body_mass_g ~ species, data = df)
cat("\nOne-Way ANOVA Summary:\n")
print(summary(anova_model))

anova_summary <- summary(anova_model)[[1]]
p_val_anova <- anova_summary[["Pr(>F)"]][1]

if (p_val_anova < 0.05) {
  cat("\nANOVA is significant. Tukey HSD Post-Hoc Test:\n")
  tukey_result <- TukeyHSD(anova_model)
  print(tukey_result)
  png("tukey_hsd_plot.png", width = 700, height = 500)
  plot(tukey_result, las = 1)
  dev.off()
} else {
  cat("\nANOVA not significant; Tukey HSD not required.\n")
}

# ============================================================
# TASK 4: NON-PARAMETRIC ANALYSIS (Kruskal-Wallis) - base R
# ============================================================

cat("\n========== KRUSKAL-WALLIS TEST: BODY MASS ~ SPECIES ==========\n")
kruskal_result <- kruskal.test(body_mass_g ~ species, data = df)
print(kruskal_result)

cat("\nComparison of ANOVA vs Kruskal-Wallis:\n")
cat("ANOVA p-value:", p_val_anova, "\n")
cat("Kruskal-Wallis p-value:", kruskal_result$p.value, "\n")
cat(ifelse((p_val_anova < 0.05) == (kruskal_result$p.value < 0.05),
           "Both tests lead to the SAME conclusion.\n",
           "The two tests lead to DIFFERENT conclusions.\n"))

# ============================================================
# TASK 5: TWO-WAY ANOVA (body_mass_g ~ species * sex) - base R
# ============================================================

cat("\n========== TWO-WAY ANOVA: BODY MASS ~ SPECIES * SEX ==========\n")
twoway_model <- aov(body_mass_g ~ species * sex, data = df)
print(summary(twoway_model))

png("interaction_plot.png", width = 700, height = 500)
interaction.plot(df$species, df$sex, df$body_mass_g,
                  col = c("blue", "red"), lwd = 2, type = "b", pch = c(16, 18),
                  xlab = "Species", ylab = "Mean Body Mass (g)",
                  trace.label = "Sex", main = "Interaction Plot: Species x Sex")
dev.off()

# ============================================================
# TASK 6: ADDITIONAL ANALYSIS - FLIPPER LENGTH
# ============================================================

cat("\n========== FLIPPER LENGTH ANALYSIS ACROSS SPECIES ==========\n")

flipper_desc <- df %>%
  group_by(species) %>%
  summarise(Mean = mean(flipper_length_mm), Median = median(flipper_length_mm),
            SD = sd(flipper_length_mm), IQR = IQR(flipper_length_mm))
print(flipper_desc)

cat("\nShapiro-Wilk Test per Species (Flipper Length):\n")
df %>% group_by(species) %>%
  summarise(p_value = shapiro.test(flipper_length_mm)$p.value) %>% print()

cat("\nLevene's Test (Flipper Length, manual):\n")
print(summary(manual_levene(df$flipper_length_mm, df$species)))

flipper_anova <- aov(flipper_length_mm ~ species, data = df)
cat("\nOne-Way ANOVA - Flipper Length ~ Species:\n")
print(summary(flipper_anova))

flipper_p <- summary(flipper_anova)[[1]][["Pr(>F)"]][1]
if (flipper_p < 0.05) {
  cat("\nTukey HSD - Flipper Length:\n")
  print(TukeyHSD(flipper_anova))
}

cat("\nKruskal-Wallis Test - Flipper Length ~ Species:\n")
print(kruskal.test(flipper_length_mm ~ species, data = df))

ggplot(df, aes(x = species, y = flipper_length_mm, fill = species)) +
  geom_boxplot() +
  labs(title = "Species-wise Flipper Length Comparison",
       x = "Species", y = "Flipper Length (mm)") +
  theme_minimal()
ggsave("boxplot_species_flipper.png", width = 7, height = 5)

# ============================================================
# TASK 7: GROUP COMPARISON PLOT (no ggpubr needed)
# ============================================================

anova_p_label <- paste0("ANOVA p-value = ", signif(p_val_anova, 4))

ggplot(df, aes(x = species, y = body_mass_g, fill = species)) +
  geom_boxplot() +
  annotate("text", x = 2, y = max(df$body_mass_g) + 300,
           label = anova_p_label, size = 4) +
  labs(title = "Group Comparison: Body Mass Across Species",
       x = "Species", y = "Body Mass (g)") +
  theme_minimal()
ggsave("group_comparison_plot.png", width = 7, height = 5)

cat("\n============================================================\n")
cat("ANALYSIS COMPLETE. Plots saved as PNG files.\n")
cat("============================================================\n")
```
