```r
# Cell 1: Creating and Merging Lists
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")
list1
list2
list12 = c(list1, list2)
list12
```

```r
# Cell 8: Vector Indexing - basics
x = 1:10
x
x[(x > 5)]
x[(x %% 2 == 0)]   # values for which x mod 2 is 0
x[(x %% 2 == 1)]   # values for which x mod 2 is 1
```

```r
# Cell 15: Factors - basic numeric coding
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
x
y = as.factor(x)
y
```

```r
# Cell 16: Factors - character coding
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x
```

```r
# Cell 17: Factors - dice example with labels
y = c(1, 4, 3, 5, 4, 2, 4)
y
possible.dieface = c(1, 2, 3, 4, 5, 6)
possible.dieface
labels.dieface = c("one", "two", "three", "four", "five", "six")
labels.dieface
facy = factor(y, levels = possible.dieface, labels = labels.dieface)
facy
```

```r
# Cell 18: class() function
class(9)
class("9")
class(print)
x = matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
x
```

```r
# Cell 19: unclass() function - brands example
brands = c("A", "A", "B", "B", "B", "B", "C")
brands
brands_fac = factor(brands)
brands_fac
unclass(brands_fac)

colours = c("blue", "green", "red")
colours
colours[unclass(brands_fac)]
```

```r
# Cell 20: unclass() - drinks factor
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x
unclass(x)
```

```r
# Cell 21: Factor with custom level order
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"),
           levels = c("water", "juice", "lemonade"))
x
unclass(x)
levels(x)
```

```r
# Cell 22: Ordered factor
income = ordered(c("high", "high", "low", "medium", "medium"),
                  levels = c("low", "medium", "high"))
income
unclass(income)
```
