## Lecture 13: Matrix Basics — Creation, Naming, Diagonal, Transpose, Sums, Means

```r
# Create a matrix
x = matrix(nrow=4, ncol=3, data=c(1:12))
x
```

```r
# Rename row and column names
rownames(x) = c("r1", "r2", "r3", "r4")
colnames(x) = c("c1", "c2", "c3")
x
```

```r
# Assign a specified number to all matrix elements
x = matrix(nrow=4, ncol=2, data=2)
x
```

```r
# Diagonal matrix -- identity matrix of dimension 3
d = diag(1, nrow=3, ncol=3)
d
```

```r
# Diagonal matrix with all numbers as 5, dimension 3
d = diag(5, nrow=3, ncol=3)
d
```

```r
# Transpose of a matrix
x = matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)
x
```

```r
xt = t(x)
xt
```

```r
# Row and column sums
x = matrix(nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8))
x
```

```r
rowSums(x)
colSums(x)
```

```r
# Row and column means
rowMeans(x)
colMeans(x)
```

## Lecture 15: Scalar Multiplication, Matrix Multiplication, Crossprod, Concatenation, Inverse, Eigen

```r
# Example: scalar multiplication and combos
x = matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)
x
```

```r
4 * x
```

```r
x + 4*x
```

```r
4*x - x
```

```r
# Multiplication of matrices with %*%
x = matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)
y = matrix(nrow=2, ncol=4, data=11:18, byrow=TRUE)
x
y
```

```r
x %*% y
```

```r
y %*% x
```

```r
# Another example: X'X and XX'
x = matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)
t(x)
```

```r
t(x) %*% x
```

```r
x %*% t(x)
```

```r
# Cross product X'X -- faster than t(x) %*% x
crossprod(x)
```

```r
# Concatenating matrices: rbind (row-wise), cbind (column-wise)
x = matrix(nrow=3, ncol=2, data=1:6, byrow=TRUE)
y = matrix(nrow=3, ncol=2, data=11:16, byrow=TRUE)
x
y
```

```r
rbind(x, y)
```

```r
cbind(x, y)
```

```r
# Inverse of a matrix with solve()
y = matrix(nrow=2, ncol=2, byrow=TRUE, data=c(84,100,100,120))
y
```

```r
solve(y)
```

```r
# Eigenvalues and eigenvectors with eigen()
eigen(y)
```

## Lecture 16: Logical Operators — `|` vs `||`, `&` vs `&&`, Logical Indexing

```r
# Use of | and ||  (single value)
x = 8
(x < 10) || (x < 2)
```

```r
x = 18
(x < 10) || (x < 2)
```

```r
# Use of | and || (vector) -- || only checks the first element
x = c(8, 18)
(x < 10) || (x < 2)
```

```r
# | operates element-wise on all elements
(x < 10) | (x < 2)
```

```r
# Use of & and && (single value)
x = 5
(x < 10) && (x > 2)
```

```r
x = 15
(x < 10) && (x > 2)
```

```r
# Use of & and && (vector) -- && only checks the first element
x = c(8, 18)
(x < 10) && (x > 2)
```

```r
# & operates element-wise on all elements
(x < 10) & (x > 2)
```

```r
# Logical indexing example: values > 2 and < 5
x = 1:6
x
```

```r
(x > 2) & (x < 5)
```

```r
x[(x > 2) & (x < 5)]
```

```r
# Logical indexing example: values > 2 or < 5
(x > 2) | (x < 5)
```

```r
x[(x > 2) | (x < 5)]
```

```r
# && only examines the FIRST element of each vector
x = 1:6
(x > 2) && (x < 5)
```

```r
# This is equivalent to the above
(x[1] > 2) & (x[1] < 5)
```

## Lecture 17: Relational Operators, Standard Logical Ops, is.logical, isTRUE/isFALSE

```r
# Standard logical operations: AND, OR, NOT
x = TRUE
y = FALSE
x & y   # x AND y
```

```r
x | y   # x OR y
```

```r
!x      # negation of x
```

```r
# is.logical() checks
x = 5
Logical1 = (x > 2)
Logical1
is.logical(Logical1)
```

```r
Logical2 = (x < 10)
Logical2
is.logical(Logical2)
```

```r
Logical3 = (x != 5)
Logical3
is.logical(Logical3)
```

```r
Logical4 = (2*x > 11)
Logical4
is.logical(Logical4)
```

```r
Logical5 = (3*x < 20)
Logical5
is.logical(Logical5)
```

```r
# Relational operators: >, <, >=, <=
8 > 7
7 < 5
7 > 7
7 >= 7
8 < 8
8 <= 8
```

```r
# Relational operators: !=, ==, !
8 != 9
9 != 9
7 == 7
7 == 8
```

```r
x = TRUE
!x
```

```r
# Vectorized relational comparisons
x = c(1, 2, 3)
y = c(4, 5, 6)
x > y   # Compares 1>4, 2>5, 3>6
```

```r
x < y   # Compares 1<4, 2<5, 3<6
```

```r
x != y  # Compares 1!=4, 2!=5, 3!=6
```

```r
x == y  # Compares 1==4, 2==5, 3==6
```

```r
# isTRUE() / isFALSE()
isTRUE(8 < 6)
```

```r
isTRUE(8 > 6)
```

```r
isFALSE(5 < 8)
```

```r
isFALSE(5 > 8)
```
