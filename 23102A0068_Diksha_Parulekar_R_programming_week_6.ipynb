```r
# Example 3: Count values satisfying a condition

x <- c(2, 4, 6, 8, 10, 12)

excount <- function(x) {
  count <- 0

  for (xval in x) {
    if (xval / 2 > 3) {
      count <- count + 1
    }
  }

  print(count)
}

excount(x)
```

```r
# Example 4: Nested for loop

child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

```r
# Example: break

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {

  if (x == "tea") {
    break
  }

  print(x)
}
```

```r
# Example: next - skip lemonade

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {

  if (x == "lemonade") {
    next
  }

  print(x)
}
```

```r
# While loop - Example 1

i <- 1

while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

```r
# While loop - Example 2

sumfunction <- function() {

  sum <- 0

  number <- as.integer(
    readline(prompt = "Please select any number less than 25: ")
  )

  while (number <= 25) {
    sum <- sum + number
    number <- number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop:",
    sum
  ))
}

sumfunction()
```

```r
# Colab-safe version

sumfunction <- function(number) {

  sum <- 0

  while (number <= 25) {
    sum <- sum + number
    number <- number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop:",
    sum
  ))
}

sumfunction(22)
```

```r
# Repeat loop - Example 3

i <- 1

repeat {

  print(i^2)

  i <- i + 2

  if (i > 10) {
    break
  }
}
```

```r
# Repeat loop - Example 4

i <- 1

repeat {

  i <- i + 1

  if (i < 10) {
    next
  }

  print(i^2)

  if (i >= 13) {
    break
  }
}
```

```r
# Function with one variable

abc <- function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
```

```r
# Function with two variables

abc <- function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

```r
# Function using sin and cos

abc <- function(x) {
  sin(x)^2 + cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
```

```r
# Function without argument

abc <- function() {

  for (i in 1:3) {
    print(i^3)
  }
}

abc()
```
