```r

# PART 1: MISSING DATA HANDLING
cat("\n===========================================================\n")
cat("PART 1: MISSING DATA HANDLING\n")
cat("============================================================\n")



# 1. Detecting Missing Values using is.na()


cat("\n--- 1. Detecting Missing Values ---\n")

x <- NA

print(x)

print(is.na(x))
```

```r
# 2. Checking Missing Values in a Vector

cat("\n--- 2. Missing Values in a Vector ---\n")

x <- c(11, NA, 13, NA)

print(x)

print(is.na(x))
```

```r
# 3. Working with Missing Data and mean()

cat("\n--- 3. Working with Missing Data ---\n")

x <- c(11, NA, 13, NA)

print(x)

cat("Mean without removing NA:\n")
print(mean(x))

cat("Mean after removing NA values:\n")
print(mean(x, na.rm = TRUE))
```

```r
# 4. NA versus NULL

cat("\n--- 4. NA versus NULL ---\n")

x <- NA

y <- NULL

cat("x = ")
print(x)

cat("y = ")
print(y)

cat("Is x NA? ")
print(is.na(x))

cat("Is y NULL? ")
print(is.null(y))
```

```r
# 5. Finding Location of Missing Values using which()

cat("\n--- 5. Location of Missing Values ---\n")

x <- c(11, NA, 13, NA)

print(x)

missing_positions <- which(is.na(x))

cat("Positions of missing values:\n")
print(missing_positions)
```

```r
# 6. Counting Number of Missing Values

cat("\n--- 6. Counting Missing Values ---\n")

x <- c(11, NA, 13, NA)

print(x)

number_of_missing <- sum(is.na(x))

cat("Number of missing values:\n")
print(number_of_missing)
```

```r
# 7. Finding Complete Cases

cat("\n--- 7. Finding Complete Cases ---\n")

x <- c(11, NA, 13, NA)

print(x)

complete <- complete.cases(x)

cat("Complete cases:\n")
print(complete)
```

```r
# 8. Removing Missing Values using na.omit()
cat("\n--- 8. Removing Missing Values using na.omit() ---\n")

x <- c(11, NA, 13, NA)

print(x)

y <- na.omit(x)

cat("After removing missing values:\n")
print(y)

cat("Mean of original vector:\n")
print(mean(x))

cat("Mean after removing NA:\n")
print(mean(y))
```

```r
# PART 6: SWITCH()

cat("\n============================================================\n")
cat("PART 6: SWITCH()\n")
cat("============================================================\n")


# 21. switch() Example 1 - Integer

cat("\n--- 21. switch() Example 1 - Integer ---\n")

result1 <- switch(
  2,
  "apple",
  "banana",
  "orange"
)

print(result1)


result2 <- switch(
  1,
  "apple",
  "banana",
  "orange"
)

print(result2)
```

```r
# 24. which() Example 4

cat("\n--- 24. which() Example 4 ---\n")

x <- c(10, 15, 8, 14, 6, 12)

print(x)

cat("Position where x == 14:\n")
print(which(x == 14))

cat("Positions where x != 12:\n")
print(which(x != 12))

cat("Positions where x > 10:\n")
print(which(x > 10))


# 25. which.min() and which.max()

cat("\n--- 25. which.min() and which.max() ---\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

print(x)

cat("Position of minimum value:\n")
print(which.min(x))

cat("Position of maximum value:\n")
print(which.max(x))

```

```r
# 26. which() with Matrix

cat("\n--- 26. which() with Matrix ---\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

print(x)

cat("Positions of odd values:\n")
print(which(x %% 2 == 1))

cat("Row and column positions of odd values:\n")
print(which(x %% 2 == 1, arr.ind = TRUE))
```
